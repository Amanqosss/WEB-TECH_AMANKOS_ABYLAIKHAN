```python
import pandas as pd
```

```python
df = pd.read_csv("sales_messy.csv")
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
print(df["country"].unique())
```

```python
print("Before:", df.shape)
```

```python
df = df.drop_duplicates()
```

```python
df = df.dropna(subset=["discount", "unit_price", "customer_id"])
```

```python
df["country"] = df["country"].str.strip().str.title()
```

```python
print("After:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
customers = pd.read_csv("customers.csv")
```

```python
print(customers["customer_id"].is_unique)
```

```python
df["customer_id"] = df["customer_id"].astype(int)
```

```python
customers["customer_id"] = customers["customer_id"].astype(int)
```

```python
before = df.shape[0]
```

```python
df = df.merge(customers, on="customer_id", how="left")
```

```python
assert df.shape[0] == before, f"Row count changed! {before} -> {df.shape[0]}"
```

```python
print("Merge OK, rows:", df.shape[0])
```

```python
df["revenue"] = df["quantity"] * df["unit_price"] * (1 - df["discount"])
```

```python
by_category = df.groupby("category")["revenue"].sum().sort_values(ascending=False)
```

```python
by_category = df.groupby("category")["revenue"].sum().sort_values(ascending=False)
print("By category:\n", by_category)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["month"] = df["order_date"].dt.to_period("M")
```

```python
by_month = df.groupby("month")["revenue"].sum().sort_values(ascending=False)
```

```python
print("\nBy month:\n", by_month)
```

```python
by_segment = df.groupby("segment")["revenue"].sum().sort_values(ascending=False)
```

```python
print("\nBy segment:\n", by_segment)
```

```python
print("\nTotal revenue:", df["revenue"].sum().round(2))
```

```python
print("Category sum:", by_category.sum().round(2))
```

```python
print("Month sum:", by_month.sum().round(2))
```

```python
print("Segment sum:", by_segment.sum().round(2))
```

```python
total = df["revenue"].sum()
```

```python
top_category = by_category.index[0]
```

```python
top_category_pct = by_category.iloc[0] / total * 100
```

```python
top_segment = by_segment.index[0]
```

```python
top_segment_pct = by_segment.iloc[0] / total * 100
```

```python
top_month = by_month.index[0]
```

```python
top_month_pct = by_month.iloc[0] / total * 100
```

```python
print(f"1. Top category: {top_category} — {top_category_pct:.1f}% of total revenue")
```

```python
print(f"2. Top segment: {top_segment} — {top_segment_pct:.1f}% of total revenue")
```

```python
print(f"3. Top month: {top_month} — {top_month_pct:.1f}% of total revenue")
```
